# What is Data Visualization?

- Data visualization is the `graphical representation` of information and data. 
- By using visual elements like charts, graphs, and maps, data visualization tools provide an accessible way to see and understand trends, outliers, and patterns in data.

## Important Libraries for Data Visualization in Python

- **Matplotlib**: Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python.
- **Seaborn**: Seaborn is a Python data visualization library based on matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics.

- **Wordcloud**: Wordcloud is a tool for making beautiful word clouds.

## ***Grammer of Graphics for plotting in Python***

- **Data**: The data to be visualized.
- **Aesthetics**: The visual properties of the data points, such as position, size, shape, color, and transparency.
- **Geometry**: The geometric shapes used to represent the data points, such as bars, lines, and points.
- **Facets**: The arrangement of the data into subsets, such as rows, columns, or panels.
- **Statistics**: The statistical transformations applied to the data, such as aggregations, smoothing, and modeling.
- **Scales**: The mapping of data values to visual properties, such as axis limits, tick marks, and color scales.
- **Themes**: The overall visual appearance of the plot, such as colors, fonts, and gridlines.
- **Guides**: The visual elements used to assist in reading the plot, such as legends, axes, and labels.
- **Interactivity**: The interactive features of the plot, such as tooltips, zooming, and panning.
- **Annotations**: The textual and graphical annotations used to highlight specific aspects of the plot.
- **Extensions**: The additional features and plugins that enhance the plot, such as animations, sliders, and widgets.

## **Parts of a plot**

- **Title**: The title of the plot, which describes the main purpose or content of the visualization.
- **Axis Labels**: The labels for the x and y axes, which describe the variables being plotted.
- **Tick Marks**: The marks along the axes that indicate the scale or range of the data.
- **Gridlines**: The lines that extend from the tick marks to assist in reading the plot.
- **Legend**: The key that explains the colors, shapes, or sizes used to represent different groups or variables.
- **Data Points**: The individual points, bars, lines, or other shapes that represent the data values.
- **Data Labels**: The text or numbers that provide additional information about the data points.
- **Annotations**: The additional text or shapes that highlight specific aspects of the plot.
- **Background**: The overall appearance of the plot, including colors, fonts, and borders.

# **Step 2: Check & Handle Missing Values (isnull) & Duplicates**

```python
# Drop rows where payment method or location zones are missing
df_clean = taxis.dropna(subset=['payment', 'pickup_zone', 'dropoff_zone']).copy()
```

# **Step 3: Check and Fix Data Types**


```python
# Check column types
print(df_clean[['pickup', 'dropoff']].dtypes)
```

```python
# you can easily create useful new columns for plotting:
# Extract hour of day and day of week
df_clean['pickup_hour'] = df_clean['pickup'].dt.hour
df_clean['pickup_day'] = df_clean['pickup'].dt.day_name()

# Calculate trip duration in minutes
df_clean['trip_duration_min'] = (df_clean['dropoff'] - df_clean['pickup']).dt.total_seconds() / 60
```

# **Step 4: Handling Outliers and Anomalies**

### **Before visualizing our data, we need to remove logical errors (like zero distances or negative fares) and extreme numerical outliers that could skew our plots.**

```python
# 1. Calculate trip duration in minutes for logical checks
df_clean['trip_duration_min'] = (df_clean['dropoff'] - df_clean['pickup']).dt.total_seconds() / 60

# 2. Filter out impossible or zero values and cap long trips at 3 hours
df_clean = df_clean[(df_clean['distance'] > 0) & 
                    (df_clean['fare'] > 0) & 
                    (df_clean['trip_duration_min'] > 0) & 
                    (df_clean['trip_duration_min'] <= 180)]

# 3. Filter extreme total price outliers using the Interquartile Range (IQR) method
Q1 = df_clean['total'].quantile(0.25)
Q3 = df_clean['total'].quantile(0.75)
IQR = Q3 - Q1

# Keep only values within 1.5 * IQR above the 75th percentile
df_clean = df_clean[df_clean['total'] <= (Q3 + 1.5 * IQR)]

# 4. Check final shape of the cleaned dataset
print(f"Cleaned dataset shape: {df_clean.shape}")
```

### **What this step did:**
- **Removed zero/negative errors:** Eliminated trips with $0 fare, 0 distance, or 0 duration that distort averages.
- **Capped unrealistic trips:** Filtered out trip durations over 180 minutes (3 hours).
- **Handled extreme fares:** Used the Interquartile Range (IQR) threshold to strip away huge fare spikes, making distributions and scatter plots much cleaner to interpret.

# **Step 8: Bivariate Analysis & Correlation Matrix**

```python

# Set global visual theme
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Scatterplot: Distance vs Total Fare
sns.scatterplot(
    data=df_clean, 
    x='distance', 
    y='total', 
    hue='color', 
    alpha=0.5, 
    ax=axes[0],
    palette={'yellow': '#f1c40f', 'green': '#2ecc71'}
)
axes[0].set_title('Distance vs. Total Fare', fontweight='bold')
axes[0].set_xlabel('Distance (miles)')
axes[0].set_ylabel('Total Fare ($)')

# 2. Regplot/Scatterplot: Trip Duration vs Total Fare
sns.regplot(
    data=df_clean, 
    x='trip_duration_min', 
    y='total', 
    ax=axes[1], 
    scatter_kws={'alpha': 0.3}, 
    line_kws={'color': 'red'}
)
axes[1].set_title('Trip Duration vs. Total Fare', fontweight='bold')
axes[1].set_xlabel('Trip Duration (minutes)')
axes[1].set_ylabel('Total Fare ($)')

# 3. Heatmap: Correlation Matrix of Numeric Variables
numeric_cols = ['passengers', 'distance', 'fare', 'tip', 'tolls', 'total', 'trip_duration_min']
corr = df_clean[numeric_cols].corr()

sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', ax=axes[2], cbar=True)
axes[2].set_title('Correlation Heatmap', fontweight='bold')

plt.tight_layout()
plt.show()
```

## **Quick Summary:** Strong relationships exist between distance, duration, and fare, while tips behave differently depending on the ride cost and payment method.

## **Distance vs. Total Fare:** There is a direct, positive correlation—as trip distance increases, total fare increases steadily. The spread widens slightly for longer trips due to variable toll charges and tips.

## **Duration vs. Total Fare:** Longer trips generally cost more, but heavy traffic causes steep fare increases even for short distances due to meter waiting rates.

## **Correlation Heatmap:** Distance, fare, and total cost have high positive correlation numbers (near +0.8 to +0.9), confirming they move closely together. In contrast, passenger count has virtually zero correlation with fare or distance.

## **Fare vs. Tip by Payment Method:** Credit card users show a steady increase in tipping as the total fare goes up, whereas cash tips remain virtually unrecorded ($0.00 logged in the system).

## 📊 Step 2: What the Pictures Taught Us!

## ⏰ 1. When do people ride taxis?

* **Busy Hours:** Most people take taxis between **4:00 PM and 7:00 PM** when they are coming home from work or school.
* **Busy Days:** **Thursdays and Fridays** are the busiest days of the week!
* **Quiet Times:** Almost nobody takes a taxi between **2:00 AM and 5:00 AM** because most people are fast asleep!

---

## 💵 2. How does the money work?

* **Farther = Costlier:** The **farther** you travel, the **more money** you pay. It’s as simple as that!
* **Traffic Costs Extra:** Sitting stuck in traffic makes the meter keep ticking, so even short rides can get expensive if the roads are jammed.
* **Airport Trips:** Going to the airport (in **Queens**) costs the most money out of all trips.

---

## 💳 3. How do people pay and tip?

* **Credit Cards:** When people pay with a credit card, they usually add a nice tip (**$2 to $5 extra**) for the driver.
* **Cash Secret:** The computer system logged cash tips as **$0**, but that just means the passenger handed cash directly to the driver, so the machine couldn't see it!

---

## 🚕 4. Yellow Taxis vs. Green Taxis

* **Yellow Taxis:** They are the kings of the city! They do the **longest trips**, go to the **airports**, and make the **most money**.
* **Green Taxis:** They stay mostly in smaller neighborhoods and do **shorter, local trips**.
